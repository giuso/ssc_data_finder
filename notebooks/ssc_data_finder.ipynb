# ssc_data_finder

Search for published papers using single-cell RNA-seq on a topic, and cross-reference them against datasets deposited in GEO.

All the search/cross-reference logic lives in the `ssc_data_finder` package (`core.py`, `europepmc.py`, `geo.py`) — this notebook is just the control surface: set a topic, run the search, inspect/filter/export the results.

```python
import pandas as pd

from ssc_data_finder.core import run_search
```

## Run a search

Set `topic` and `max_results`, then run the cell below.

```python
topic = "systemic sclerosis"
max_results = 25

rows = run_search(topic, max_results=max_results)
df = pd.DataFrame(rows)
print(f"{len(df)} papers found for '{topic}'")
df
```
